```python
#melihat ringkasan statistik deskriptif
df.describe()
```

```python
#distribusi target
df['Target'].value_counts()
```

Mengetahui jumlah Dropout, Enrolled, Graduate dan
Mengecek imbalance data


```python
#visualisasi target
df['Target'].value_counts().plot(kind='bar')
plt.title("Distribusi Status Mahasiswa")
plt.show()
```

visual melihat dominasi kelas dan membantu evaluasi model

```python
#identifikasi fitur numerik (Memisahkan fitur numerik untuk analisis lanjutan (korelasi, scaling))
numerical_cols = df.select_dtypes(include=['int64','float64']).columns
```

```python
#histogram nilai akademik
df['Curricular units 1st sem (grade)'].hist()
plt.show()
```

melihat distribusi nilai dan mendeteksi skewed data

```python
#boxplot (outlier)
sns.boxplot(x=df['Curricular units 2nd sem (grade)'])
plt.show()
```

mengetahui apakah ada nilai extrem dan outlier dicatat, bukan langsung dihapus

```python
#nilai vs status mahasiswa
sns.boxplot(x='Target', y='Curricular units 1st sem (grade)', data=df)
plt.show()
```

melihat pengaruh nilai terhadap dropout/graduate dan validasi hubungan logis data

```python
#faktor finansial vs target
sns.countplot(x='Scholarship holder', hue='Target', data=df)
plt.show()
```

melihat pengaruh beasiswa terhadap status mahasiswa

```python
sns.boxplot(
    x='Tuition fees up to date',
    y='Curricular units 2nd sem (grade)',
    data=df
)
plt.title("Pengaruh Pembayaran Biaya Kuliah")
plt.show()
```

```python
sns.boxplot(
    x='Target',
    y='Curricular units 2nd sem (grade)',
    data=df
)
plt.title("Pengaruh Beasiswa terhadap Prestasi")
plt.show()
```

```python
#hitung korelasi
corr = df[numerical_cols].corr()
```

mengetahui hubungan antar fitur numerik

```python
sns.boxplot(data=df[['Curricular units 1st sem (grade)',
                     'Curricular units 2nd sem (grade)']])
plt.title("Perbandingan Nilai Semester 1 dan 2")
plt.show()
```

```python
df['Delta_Grade'] = (
    df['Curricular units 2nd sem (grade)'] -
    df['Curricular units 1st sem (grade)']
)
```

```python
sns.histplot(df['Delta_Grade'], bins=30, kde=True)
plt.title("Distribusi Perubahan Prestasi Mahasiswa")
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64'])
corr_matrix = numeric_cols.corr()
```

```python
#heatmap korelasi
sns.heatmap(corr, cmap='coolwarm')
plt.show()
```
